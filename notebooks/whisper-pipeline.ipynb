```python
from openai import OpenAI, APIConnectionError, APIStatusError
```

```python
WHISPER_BASE_URL = "https://whisper-large-v3-turbo-quantized-llama-serve.apps.cluster-m2r54.m2r54.sandbox2855.opentlc.com/v1"
WHISPER_MODEL = "whisper-large-v3-turbo-quantized"
WHISPER_API_KEY = "ollama"

GRANITE_BASE_URL="https://granite32-8b-llama-serve.apps.cluster-m2r54.m2r54.sandbox2855.opentlc.com/v1"
```

```python
file = open("audio/audio1.mp3", "rb")
content =  file.read()
```

```python
whisper_client = OpenAI(base_url=WHISPER_BASE_URL, api_key=WHISPER_API_KEY)

kwargs = {
        "model": WHISPER_MODEL,
        "file": ("audio", file),
        "response_format": "json",
        "prompt": "Separate text into timestamped segments",
        "timestamp_granularities": "segment"
    }

result = whisper_client.audio.transcriptions.create(**kwargs)
print(f"Received transcription result: {result}")
```

```python
granite_client = OpenAI(
    base_url=GRANITE_BASE_URL,
    api_key="YOUR_TOKEN"  # often a bearer token from OpenShift
)

response = granite_client.chat.completions.create(
    model="granite32-8b",
    messages=[
        {
            "role": "user",
            "content": f"Parse this transcript and return a formatted transcript between Pilot and Control Tower indicating who said what. Example:\nPilot: getting ready for takeoff\nController: sounds good let us know what happens.\n\nTranscript:\n\n{result}"
        }],
    temperature=0.3
)

print(response.choices[0].message.content)
```
